```python
# %%
import time

print("演示 time.perf_counter() 的基本用法")

# 记录循环开始时间
start = time.perf_counter()
print(f"开始时间: {start:.6f} 秒")
# 模拟计算任务
total = 0
for i in range(1_000_000):
    total += i**0.5

# 计算耗时
elapsed = time.perf_counter() - start

print(f"计算耗时: {elapsed:.6f} 秒")
```

```python
import sys, time
sys.path.insert(0, '../')
from robots.fanuc.fanuc import Fanuc

robot = Fanuc(host="172.30.109.22", port=16001, speed=150, term_type="FINE", term_value=0)
robot.connect()

obs = robot.get_observation()
x, y, z, w, p, r = obs["position"]
print(f"初始位姿: {obs['position']}")

hz = 10
dt = 1.0 / hz
waypoints = [
    (x + 20, y, z, w, p, r),
    (x,      y, z, w, p, r),
] * 5  # 来回 5 次，共 10 步

for i, pos in enumerate(waypoints):
    t0 = time.perf_counter()

    obs = robot.get_observation()
    action = {"position": pos, "speed": 150, "term_type": "FINE", "term_value": 0}
    fut = robot.send_action(action)
    err = fut.result(timeout=10.0)

    elapsed = time.perf_counter() - t0
    print(f"[{i:02d}] pos={obs['position'][:3]} -> target={pos[:3]} err={err} elapsed={elapsed:.3f}s")
    time.sleep(max(0, dt - elapsed))

robot.disconnect()
```
